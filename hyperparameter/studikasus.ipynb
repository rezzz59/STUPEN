```python
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

X, y = fetch_california_housing(return_X_y=True)

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

print("Shape of training data ", X_train.shape)
print("Shape of test data ", X_test.shape)
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error

#inisialisasi model random forest regressor
rf = RandomForestRegressor(random_state=42)
rf.fit(X_train, y_train)

y_pred = rf.predict(X_test)
mse = mean_squared_error(y_test, y_pred)
print(f"Mean Squared Error tanpa tuning : {mse:.2f}")
```

```python
import time
import pandas as pd
import numpy as np
from sklearn.model_selection import RandomizedSearchCV

start_time = time.time()

param_distributions = {
    'n_estimators' : np.arange(100, 500, 100),
    'max_depth' : [None] + list(np.arange(10, 50, 10)),
    'min_samples_split' : np.arange(2, 11, 2),
    'min_samples_leaf' : np.arange(1,5),
    'bootstrap' : [True, False]
}

grid_search = RandomizedSearchCV(estimator=rf, param_distributions=param_distributions, n_iter = 5, cv=3, n_jobs=-1, verbose=2)
grid_search.fit(X_train, y_train)

print(f"Best parameter (Grid Search) : {grid_search.best_params_}")
best_rf_random = grid_search.best_estimator_

y_pred_random = best_rf_random.predict(X_test)
mse_random = mean_squared_error(y_test, y_pred_random)
print(f"MSE After Random Search : {mse_random:.2f}")
end_time = time.time()
excution_time = end_time - start_time
print("waktu eksekusi : ", excution_time)
```

```python
from skopt import BayesSearchCV


start_time = time.time()
param_space = {
    'n_estimators' : (100, 500),
    'max_depth' : (10, 50),
    'min_samples_split' : (2, 10),
    'min_samples_leaf' : (1, 4),
    'bootstrap' : [True, False]
}

bayes_search = BayesSearchCV(estimator=rf, search_spaces=param_space, n_iter=32, cv=3, n_jobs=1, verbose=2, random_state=42)
bayes_search.fit(X_train, y_train)

print("Best parameter dari hasil hyperparameter tuning : ", bayes_search.best_params_)
best_bayes_search = bayes_search.best_estimator_

y_pred_bayes = best_bayes_search.predict(X_test)
mse_bayes = mean_squared_error(y_test, y_pred_bayes)
print(f'MSE after bayes search : {mse_bayes:.2f}')
end_time = time.time()
excution_time = end_time - start_time
print("waktu eksekusi bayes_search : ", excution_time)
```
